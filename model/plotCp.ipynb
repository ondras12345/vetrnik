```python
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import math
```

```python
df=pd.read_csv("windCp.csv")
df
```

```python
ax = plt.axes()
for b in df.beta.unique():
    tmp = df[df.beta == b]
    line = ax.plot(tmp["lambda"], tmp.Cp, label=f"beta = {b*180/math.pi:.1f} deg")
ax.legend()
plt.ylim(-0.2, 0.5)
plt.grid()
```

```python
matplotlib.use("pgf")
matplotlib.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'serif',
    'text.usetex': True,
    'pgf.rcfonts': False,
})
```
